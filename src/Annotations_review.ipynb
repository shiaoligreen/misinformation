```python
import json
import os
import pandas as pd
```

```python
annotations_path = os.path.join('..', 'data', 'preprocessed', 'consolidated_annotations.json')

annotations = pd.read_json(annotations_path)

# split and grab text before url
annotations['text_before_url'] = annotations['Text'].str.split(r'https?://|www\.', regex=True).str[0].str.strip()

# drop duplicates based on new column and keep the first instance
annotations_cleaned = annotations.drop_duplicates(subset=['text_before_url'], keep='first')

# look at the change in size
print(f"Original row count: {len(annotations)}")
print(f"Cleaned row count: {len(annotations_cleaned)}")
```

```python
# export updated file as v2
new_file_path = os.path.join('..', 'data', 'preprocessed', 'consolidated_annotations_v2.json')
annotations_cleaned.to_json(new_file_path)
```

```python
dataset_path = os.path.join('..', 'data', 'preprocessed', 'cleaned_dataset.csv')
dataset = pd.read_csv(dataset_path, index_col=0)

# split and grab text before url
dataset['text_before_url'] = dataset['text'].str.split(r'https?://|www\.', regex=True).str[0].str.strip()

# drop duplicates based on new column and keep the first instance
dataset_cleaned = dataset.drop_duplicates(subset=['text_before_url'], keep='first')

# look at the change in size
print(f"Original row count: {len(dataset)}")
print(f"Cleaned row count: {len(dataset_cleaned)}")
```

```python
# export updated file as v2
new_file_path = os.path.join('..', 'data', 'preprocessed', 'cleaned_dataset_v2.csv')
dataset_cleaned.to_csv(new_file_path)
```
